### Choose our K-value :- Elbow Method

```python
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

```python
X, y= make_blobs(
    n_samples=1000,
    n_features=2,
    centers=4,
    random_state=42
)
```

```python
wcss=[]

for k in range(1,21):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1,21), y= wcss, marker="o")
```

### Kneed Module- 
A Python library for detecting knee (elbow) points in curves using the Kneedle algorithm.                   Given a set of x and y values, kneed returns the point of maximum curvature.

Installation - !pip install kneed

```python
from kneed import KneeLocator
```

```python
knee=KneeLocator(range(1,21), wcss, curve="convex", direction="decreasing")
```

```python
print("Optimal K =", knee.elbow)
```
